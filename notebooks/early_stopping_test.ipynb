Jalankan EarlyStopping

```python
early_stopping = EarlyStopping(
    patience=3
)

for epoch, val_loss in enumerate(val_losses):

    print(
        f"Epoch {epoch+1} "
        f"Val Loss: {val_loss}"
    )

    early_stopping(val_loss)

    if early_stopping.early_stop:

        print("STOP TRAINING")

        break
```
