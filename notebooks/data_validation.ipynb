Import Library

```python
import os
import pandas as pd
import numpy as np
from PIL import Image
```

Load Master Metadata

```python
df = pd.read_csv(
    "../data/metadata/harmonized_labels.csv"
)

print(df.shape)

df.head()
```

Cek Missing Value

```python
print(df.isnull().sum())
```

Cek Duplicate Path

```python
duplicate_count = df["image_path"].duplicated().sum()

print("Duplicate paths:", duplicate_count)
```

Cek File Exists

```python
missing_files = []

for path in df["image_path"]:
    
    if not os.path.exists(path):
        missing_files.append(path)

print("Missing files:", len(missing_files))
```

Tampilkan Missing File (Kalau Ada)

```python
missing_files[:10]
```

Cek Corrupted Image

```python
from tqdm import tqdm

corrupted_images = []

for path in tqdm(df["image_path"]):

    try:
        with Image.open(path) as img:
            img.verify()

    except Exception as e:
        corrupted_images.append(path)

print("Corrupted images:", len(corrupted_images))
```
