Import Library

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
```

Load Clean Metadata

```python
df = pd.read_csv(
    "../data/metadata/harmonized_clean.csv"
)

print(df.shape)

df.head()
```

Buat Kolom Stratifikasi

```python
df["cardiomegaly"] = (
    df["cardiomegaly"]
    .fillna(-1)
    .astype(int)
)

df["tb"] = (
    df["tb"]
    .fillna(-1)
    .astype(int)
)

df["combined_label"] = (
    df["cardiomegaly"].astype(str)
    + "_"
    + df["tb"].astype(str)
)

df["combined_label"].value_counts()
```

Split Train dan Temp

```python
#70% train
#30% sementara
train_df, temp_df = train_test_split(
    df,
    test_size=0.30,
    stratify=df["combined_label"],
    random_state=42
)
```

Split Validation dan Test

```python
#15% val
#15% test
val_df, test_df = train_test_split(
    temp_df,
    test_size=0.50,
    stratify=temp_df["combined_label"],
    random_state=42
)
```

Cek Ukuran

```python
print("Train:", train_df.shape)
print("Validation:", val_df.shape)
print("Test:", test_df.shape)
```

Cek Distribusi Label

```python
#TRAIN
print(
    train_df["combined_label"]
    .value_counts(normalize=True)
)
```

```python
#VALIDATION
print(
    val_df["combined_label"]
    .value_counts(normalize=True)
)
```

```python
#TEST
print(
    test_df["combined_label"]
    .value_counts(normalize=True)
)
```

Simpan CSV

```python
#Train
train_df.to_csv(
    "../data/train/train.csv",
    index=False
)
```

```python
#Validation
val_df.to_csv(
    "../data/val/val.csv",
    index=False
)
```

```python
#Test
test_df.to_csv(
    "../data/test/test.csv",
    index=False
)
```
